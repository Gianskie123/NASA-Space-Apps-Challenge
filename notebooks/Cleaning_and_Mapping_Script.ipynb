**MASTERLIST**

```python
species_masterlist = pd.read_csv("https://docs.google.com/spreadsheets/d/1k4oiChJPhkroGhKxU7gX2ngxLTs6Z-9jSWiCBpUINCw/export?format=csv")
```

```python
# Create a mapping from species to light tolerance
species_to_tolerance = species_masterlist.set_index('Species Name')['Light Tolerance'].to_dict()
# Map the light tolerance to df based on species name
df['Bird Type - Light Tolerance'] = df['COMMON NAME'].map(species_to_tolerance)
```

```python
# Create a mapping from species to migration status
species_to_migration = species_masterlist.set_index('Species Name')['Migratory Status'].to_dict()
# Map the light tolerance to df_2012 based on species name
df['Bird Type - Migrant/Resident'] = df['COMMON NAME'].map(species_to_migration)
```

```python
column_new_names ={
    'LOCALITY':'Site Name',
    'COMMON NAME':'Species Name',
    'SCIENTIFIC NAME':'Scientific Name',
    'OBSERVATION DATE':'Date',
    'OBSERVATION COUNT':'Bird Count',
    'LATITUDE':'Latitude',
    'LONGITUDE':'Longitude'
}

for i in column_new_names.keys():
  df.rename(columns={i:column_new_names[i]},inplace=True)
```

```python
df['Date'] = df['Date'].astype('datetime64[ns]')
```

```python
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year
df = df.drop('Date',axis=1)
```

```python
df.info()
```

**FILTERING GMA AREAS**

```python
lat_min, lat_max = 14.3, 14.8
lon_min, lon_max = 120.9, 121.15

# Filter to KEEP rows *inside* this range
df = df[
    ((df['Latitude'] >= lat_min) &
      (df['Latitude'] <= lat_max) &
      (df['Longitude'] >= lon_min) &
      (df['Longitude'] <= lon_max))
]
```

```python
df.info()
```

**REPLACING X VALUES WITH MEDIAN**

```python
df['Bird Count'] = pd.to_numeric(df['Bird Count'], errors='coerce')

# 4. Fill NaNs ('X' values) with the MEDIAN count for each Species
# transform('median') applies the median of the group back to each NaN row
df['Bird Count'] = df['Bird Count'].fillna(
    df.groupby('Species Name')['Bird Count'].transform('median')
)

# 5. Fallback (If a species only has 'X' entries and no median exists, use 1)
df['Bird Count'] = df['Bird Count'].fillna(1)

# 6. Ensure bird counts are whole numbers (integers)
df['Bird Count'] = df['Bird Count'].round().astype(int)

# Verify results
print(f"Total Rows: {len(df)}")
print(df[['Species Name', 'Bird Count']].head())
```
